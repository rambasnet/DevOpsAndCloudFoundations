# Docker Basics

- https://www.docker.com/ 
- Docker is a platform for developing, shipping, and running applications in containers
- Containers are lightweight, portable, and self-sufficient units that package an application and its dependencies together
- Docker allows developers to create, deploy, and run applications in a consistent environment across different systems
- Docker uses a client-server architecture, where the Docker client communicates with the Docker daemon to build, run, and manage containers
- Docker images are the blueprints for containers, containing the application code, libraries, and dependencies needed to run the application
- Docker containers are instances of Docker images that can be run, stopped, and managed independently
- Docker provides a command-line interface (CLI) and a graphical user interface (GUI) for managing containers, images, and other Docker resources
- Docker is widely used in DevOps and cloud computing for continuous integration and continuous deployment (CI/CD) pipelines, microservices architecture, and scalable application deployment

## Dockerfile

- A Dockerfile is a text file that contains a set of instructions to build a Docker image. 
- It defines the base image, application code, dependencies, and configuration needed to create a container.
- The Dockerfile is used by the Docker CLI to build an image using the `docker build` command.
- various instructions can be used in a Dockerfile, such as `FROM`, `RUN`, `COPY`, `WORKDIR`, `CMD`, and more, to define the image's behavior and contents.
- FROM: Specifies the base image to use for the Docker image.
- RUN: Executes commands in the image during the build process.
- COPY: Copies files or directories from the host machine to the image.
- WORKDIR: Sets the working directory for subsequent instructions in the Dockerfile.
- ARG: Defines build-time variables that can be passed to the Docker build command.
- USER: Sets the user to use when running the container.
- CMD: Specifies the command to run when the container starts.
- ENV: Sets environment variables in the image.

- Example Dockerfile:

```Dockerfile
FROM python:3.14
RUN apt update \
    && apt install -y --no-install-recommends \
    ca-certificates \
    git \
    curl \
    && rm -rf /var/lib/apt/lists/*

ARG USERNAME=devops
USER ${USERNAME}
WORKDIR ${HOME}
```

## Docker Compose
- Docker Compose is a tool for defining and running multi-container Docker applications.
- It allows you to define the services, networks, and volumes needed for your application in a single YAML file called `docker-compose.yml`.
- You can use the `docker-compose` command to build, start, stop, and manage the services defined in the `docker-compose.yml` file.
- Docker Compose simplifies the process of managing complex applications with multiple containers, making it easier to develop, test, and deploy applications in a consistent environment.
- Example `docker-compose.yml` file:

```yaml
services: # Defines the services that make up the application
  web: # Defines the web service, which is a container that runs the web application
    build: . # Builds the Docker image from the Dockerfile in the current directory
    Dockerfile: Dockerfile # Specifies the Dockerfile to use for building the image
    ports: # Maps port 8000 on the host to port 8000 in the container
      - "8000:8000"
  db: # Defines the db service, which is a container that runs the PostgreSQL database
    image: postgres:latest # Uses the latest official PostgreSQL image from Docker Hub
    environment: # Sets environment variables for the PostgreSQL container
      POSTGRES_USER: myuser
      POSTGRES_PASSWORD: mypassword
      POSTGRES_DB: mydatabase
```

- common Docker Compose commands:
- `docker-compose up -d --build`: Builds, (re)creates, starts, and attaches to containers for a service defined in the `docker-compose.yml` file. 
    - -d: run in detached mode
    - --build: rebuilds the images if necessary.
- `docker-compose down`: Stops and removes containers, networks, and volumes defined in the `docker-compose.yml` file.
