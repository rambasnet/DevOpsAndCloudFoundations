## Loops
- life is full of loops and iteration; everyday routines such as going to college, cooking food, slicing vegetables, etc.
- E.g. slicing a carrot
    1. get cutting board
    2. get knife
    3. place carrot on cutting board
    4. lift knife
    - advance carrot
    - slice carrot
    - if you see only your finger, then quit else go to step 4
- step 4-7 form a loop!

- bash provides three types of loops:
    1. for loop
    2. while loop
    3. until loop

## For loop
- syntax
```bash
for (( init; condition; update )); do 
    # body
done
```

```bash
for val in val1 val2 ...; do
    # body
done
```

````python
%%bash
# for loop similar to C++ for loop
for (( count=1; count<=20; count++ )); do
    if (( $count%2 == 0 )); then
        echo $count
    fi
done
````

````python
%%bash
# looping through a list of values
# similar to Python loop
for i in 1 2 35; do
    echo -n "$i "
done
````

````python
%%bash
# write a script to print number from 1-20
for i in {1..20}; do
    echo -n "$i " 
done
````

````python
%%bash
for w in word1 "word 2" "word3" 10; do
    echo $w
done
````

````python
%%bash

# print line and total characters in each line for contents of home dir
count=1
for file in $(ls ~); do
    echo "Line $count ($file) contains $(echo -n $file | wc -c) characters"
    (( count++ ))
done
````

## While loop
```bash
while condition; do
    #body
done
```

````python
%%bash
count=1
while (( $count <= 20 )); do
    if (( $count%2 == 0 )); then
        echo $count
        #exit 0;
    fi
    (( count++ ))
done
````
