# Shell Script

- a file containing a series of commands that can be executed by the shell
- Shell reads the script file and executes the commands in order as if they were typed at the command line
- Shell scripts are used to automate repetitive tasks, manage system configurations, and perform complex operations that would be tedious to do manually
- shell scripts include control structures, variables, and functions to create more complex and dynamic scripts
- Dev Ops need to know shell scripting to automate tasks and manage systems efficiently
- see `shellscripts/hello.sh` for a simple shell script example

## Data Types

- bash supports several data types, including strings, integers, arrays, and associative arrays

## Declaring Variables
- syntax:
 
 ```bash
 variable_name=value
 ```
 - note that there should be no spaces before or after the equal sign
 - space has special meaning in bash, so it is important to avoid spaces when declaring variables

 ## Using Variables
- to use a variable, prefix its name with a dollar sign ($)

````python
%%bash
name="John Smith"
echo "Hello $name. Nice to meet you!"
````

````python
%%bash
num=100 # int
price=999.123456789 # float
name="John Smith" # string
person='Jane Smith' # string
echo name = $name
````

## Standard Output

- the standard output (stdout) is the default destination for output from commands and programs
- commands:
    - `echo` - prints text or variables to the standard output
    - `printf` - formats and prints text or variables to the standard output

```bash
printf <format> <space separated arguments>
```

- formats:
    - %x - hex
    - %o - octal
    - %d or %i - signed decimal
    - %s - string
    - %f - float

- NOTE: echo adds a newline character at the end of the output, while printf does not.


````python
%%bash
name="John Smith"
echo "hi there! $name"
printf "float data= %.2f  and variable data " 2.3456
printf "Good bye..."
````

````python
# multiple values for single format string
! printf "%.1f\n" 1.44 2.33 3.49 
````

````python
%%bash
num=100 # int
price=999.123456789 # float
name="John Smith" # string
person='Jane Smith' # string
echo name = $name
printf "Hello %s. Nice meeting you!\n" "$person" # need quotes for values with spaces
printf 'Price of new iphone is $%.3f\n' $price
echo "Mr. ${name}" # can also use curly braces
````
