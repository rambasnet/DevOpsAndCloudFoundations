# Unittesting Bash with Bats

- unittesting is a software testing technique where individual units or components of a software application are tested in isolation to ensure that they function as expected
-  in the context of Bash scripting, unit testing can be performed using a framework called Bats (Bash Automated Testing System)
- Bats is a testing framework specifically designed for Bash scripts, allowing developers to write and execute unit tests for their Bash functions and scripts in a structured manner
- basts must be installed in the system to run the tests. It can be installed using package managers like Homebrew, apt, or yum, depending on the operating system

## Install Bats on Linux/Codespaces

```bash
sudo apt update # must be run before installing any package
sudo apt install bats
```

## Using Bats for Unit Testing
- Bats provides a simple syntax for writing test cases. Each test case is defined using the `@test` annotation, followed by a description of the test. 
- Inside the test case, you can use standard Bash commands and assertions to verify the expected behavior of your functions or scripts.
- use the `run` command to execute the function or script being tested, and then use assertions to check the output or return value against the expected result.

- syntax for writing a test case in Bats:

```bash
test_file.bats

@test "description of the test case" {
    # Test logic and assertions go here
    run some_command
    [ assertion ]
    [ assertion ]
}
```
- must run the test file using the `bats` command, followed by the path to the test file. For example:

```bash
bats path/to/test_file.bats
```

- see `demos/bats` for examples of unit testing using Bats

````python
%%bash
add() {
    echo $(( $1 + $2 ))
}
add 10 20
````

````python
%%bats

# bats unittest
@test "Adding two positive integers" {
    run add 10 20
    [ "$status" -eq 0 ]
    [ "$output" = "30" ]
}
````
