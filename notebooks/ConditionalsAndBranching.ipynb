# Conditionals
- also called logical statements that help program make decision and control the execution of commands

## One-way selection

- bash provides the `if` statement to perform one-way selection. 
- the syntax is as follows:
```bash
if a_command; then
    command1
    command2
fi
```

- if `a-command` executed successfully with exit status 0, then only execute `command1`
    - if `a-command` failed with exit status other than 0, then skip the commands and continue with the next command after `fi`

## Exit status
- commands (including bash scripts) issue a value to the system when they terminate, called the exit status
- this is an integer value in the range $0-255$ indicating success or failure
- by conventions, value of $0$ (zero) indicates success and any other value indicates failure
- `$?` stores the exit status of last command executed

````python
%%bash
ls -d /binddfds
echo $? #print exit status of last command
````

````python
%%bash
ls -d /bin/usr
echo "Exit status = $?"
````

````python
%%bash
ls -d /usr/bin
echo "Exit status = $?"
````

````python
%%bash
true #true is 0!
echo $?
````

````python
%%bash
false # false is 1!
echo $?
````

## Testing conditional expression
- bash provides three different ways to test conditional expressions

- `[ expression ]` 
    - basic test
    - =, ==, !=, -ge, -ne, -eq, -le, -lt, -gt
    - `-a` and `-o` for logical AND and OR
    - `!` for logical NOT
    - `-z` and `-n` for string length tests
        - `-z` returns true if string length is zero
        - `-n` returns true if string length is non-zero

- `(( expression ))` 
    - arithmetic/integer truth tests; must for integers!
    - <=, <, >, >=, &&, ||, !=
    -
- `[[ expression ]]` 
    - for compound expressions 
    - more useful, modern and easier version of test
    - good for evaluating file and pathnames
    - good fo arithmethic integer truth tests

````python
%%bash
#!/bin/bash
#test-string: evaluate the value of a string
answer=
if [[ -n "$answer" ]]; then # if length of string is > 0
    echo "There's some answer"
fi

if [[ -z "$answer" ]]; then # length is 0
    echo "There's no answer"
fi
````

## Two-way selection

- bash provides the `if-else` statement to perform two-way selection

- syntax:

```bash
if commands; then
	#commands...
else
	#commands...
fi
```
- if `commands` executed successfully with exit status 0, then execute the commands in the `then` block
- if `commands` failed with exit status other than 0, then execute the commands in the `else` block

````python
%%bash
#!/bin/bash
a=100
b=20
if (( $a < $b )); then
    echo "$a is less than $b"
else
    echo "$a is not less than $b"
fi
````

````python
%%bash
#!/bin/bash
# run demo-scripts/conditional.sh
# test-file: Evaluate the status/type of a file
FILE=README.md
# read -p "Enter a existing file" FILE
if [[ -e "$FILE" ]]; then
    if [[ -f "$FILE" ]]; then
        echo "$FILE is a regular file."
    fi
    if [[ -d "$FILE" ]]; then
        echo "$FILE is a directory."
    fi
    if [[ -r "$FILE" ]]; then
        echo "$FILE is readable."
    fi
    if [[ -w "$FILE" ]]; then
        echo "$FILE is writable."
    fi
    if [[ -x "$FILE" ]]; then
        echo "$FILE is executable/searchable."
    fi
else
    echo "$FILE does not exist"
    exit 1
fi
exit 0
````

## Multi-way selection

- bash provides the `if-elif-else` statement to perform multi-way selection
- syntax:

```bash
if commands; then
	#commands...
elif commands; then
	#commands...
elif commands; then
    #commands...
...

fi
```

- if `commands` executed successfully with exit status 0, then execute the commands in the `then` block
- if `commands` failed with exit status other than 0, then check the next `elif` block and so on
- if all `if` and `elif` blocks fail, then execute the commands in the `else` block


## Using multiple files in bash scripts
- bash provides the `source` command to include and execute commands from another file in the current shell
- when a script is sourced, it runs in the current shell and any changes made to the environment variables will persist after the script has finished executing
- sourced file is executed in the current shell, so any variables or functions defined in the sourced file will be available in the current shell after the script has finished executing
- you've to guard against sourcing executable code that may have side effects or modify the environment in unexpected ways

### guarding against sourcing executable code
- you can use the following code snippet to check if the script is being sourced or executed directly   
-  Only run main execution if the script is run directly, NOT when sourced

```bash
if [[ "${BASH_SOURCE[0]}" == "${0}" ]]; then
    # main execution code here
fi
```
- similar to the `if __name__ == "__main__":` idiom in Python, which is used to check if a script is being run directly or imported as a module.
