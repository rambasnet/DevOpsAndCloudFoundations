# Git Best Practices

- git is a distributed version control system that allows multiple developers to work on a project simultaneously without interfering with each other's work.
- Git is a powerful tool that can help you manage your codebase, track changes, and collaborate with others effectively. 
- However, to make the most of git, it's important to follow best practices that can help you avoid common pitfalls and ensure that your codebase remains clean and maintainable.
- Here are some best practices for using git:
1. **Use descriptive commit messages**: When committing changes, use clear and concise messages that describe the changes made. This will help you and your team understand the history of the project and make it easier to track down bugs or issues.
 2. **Commit frequently**: Commit your changes frequently to avoid losing work and to make it easier to track changes over time. Small, frequent commits are easier to review and understand than large, infrequent ones.
 3. **Use branches**: Use branches to isolate new features or bug fixes from the main codebase. This allows you to work on new features without affecting the stability of the main codebase.
 4. **Keep your branches up-to-date**: Regularly merge changes from the main branch into your feature branches to avoid conflicts and ensure that your code is always up-to-date with the latest changes.
 5. **Use pull requests**: Use pull requests to review and discuss changes before merging them into the main codebase. This allows for collaboration and ensures that code is reviewed for quality and consistency.
 6. **Avoid committing sensitive information**: Avoid committing sensitive information such as passwords or API keys to your git repository. Use environment variables or configuration files instead.
 7. **Use .gitignore**: Use a .gitignore file to exclude files that should not be tracked by git, such as build artifacts or temporary files.
 8. **Keep your repository organized**: Keep your repository organized by using a clear directory structure and naming conventions for files and folders.

 ## Using Git Branches

 - branches are a powerful feature of git that allow you to work on new features or bug fixes without affecting the main codebase. 
 - Here are some best practices for using git branches:

- **Use descriptive branch names**: Use clear and descriptive names for your branches that indicate the purpose of the branch. For example, use `feature/login-page` for a new login page feature or `bugfix/fix-login-error` for a bug fix related to login errors, `issue/issue-123` for a branch related to a specific issue number, etc.
- using descriptive branch names makes it easier to understand the purpose of the branch and helps avoid confusion when working with multiple branches
- **Keep branches small and focused**: Keep your branches small and focused on a specific feature or bug fix. This makes it easier to review and merge changes, and reduces the risk of introducing bugs or conflicts.
- **Regularly merge changes from the main branch**: Regularly merge changes from the main branch into your feature branches to avoid conflicts and ensure that your code is always up-to-date with the latest changes.
- **Always commit on branch** - Always commit your changes on the branch you are working on, rather than committing directly to the main branch. This helps keep the main branch stable and reduces the risk of introducing bugs or conflicts.

## Enforcing Proper Branch name Conventions
- you can use git hooks to enforce proper branch name conventions. 
- Git hooks are scripts that run automatically when certain git events occur, such as committing changes or pushing to a remote repository.
- must use a `pre-commit` hook to enforce branch name conventions.

### Example of a pre-commit hook to enforce branch name conventions

- create a file named `pre-commit` in the `.git/hooks` directory of your repository and add the following code to it:
```bash
#!/bin/sh
# Get the current branch name
branch_name=$(git rev-parse --abbrev-ref HEAD)
# Do not allow commit on main
if [[ "$current_branch" == "main" ]]; then
    echo "Commit on main is not allowed. Please use a feature branch."
    exit 1
fi
if ! [[ "$branch_name" =~ ^(feature|bugfix|issue)/[a-z0-9._-]+$ ]]; then
  echo "Error: Branch name '$branch_name' does not follow the naming convention."
  echo "Branch names must start with 'feature/', 'bugfix/', or 'issue/' followed by a descriptive name or number."
  exit 1
fi
exit 0
```

- change the permissions of the `pre-commit` file to make it executable:
```bash
chmod +x .git/hooks/pre-commit
```
- tell git to use the pre-commit hook by running the following command:
```bash
git config core.hooksPath .githooks
```
